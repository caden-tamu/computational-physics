```python
import matplotlib.pyplot as plt
import pandas as pd

# Load in data from the CSV file
data = pd.read_csv("population_growth_real1.0.csv")

# Plotting N(t) Growth
plt.figure(figsize=(10, 6))
plt.title("Population Growth: N(t) Numerical vs Analytical")
plt.plot(data["Time"], data["N_numeric"], "--r", label= "N(t) Numerical (Euler)")
plt.plot(data["Time"], data["N_analytical"], "-b", label= "N(t) Analytical")
plt.legend()
plt.grid()
plt.xlabel("Time (years)")
plt.ylabel("Population (millions)")


# Plotting R(t) Activity
plt.figure(figsize=(10, 6))
plt.title("Population Growth: R(t) = dN/dt Numerical vs Analytical")
plt.plot(data["Time"], data["R_numeric"], "--g", label="R(t) Numerical (Euler)")
plt.plot(data["Time"], data["R_analytical"], "-m", label="R(t) Analytical")
plt.legend()
plt.grid()
plt.xlabel("Time (years)")
plt.ylabel("Growth Rate (dN/dt, millions/year)")

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the data
data = pd.read_csv("population_growth_real_with_error.csv")

# Plot the errors in N(t)
plt.figure(figsize=(10, 6))
plt.plot(data["Time"], data["Error_N"], label="Error in N(t)", color="red")
plt.xlabel("Time (years)")
plt.ylabel("Error in N(t)")
plt.title("Numerical Error in N(t) Over Time")
plt.grid()
plt.legend()

# Plot the errors in R(t)
plt.figure(figsize=(10, 6))
plt.plot(data["Time"], data["Error_R"], label="Error in R(t)", color="blue")
plt.xlabel("Time (years)")
plt.ylabel("Error in R(t)")
plt.title("Numerical Error in R(t) Over Time")
plt.grid()
plt.legend()

# Show the plots
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd

# Read CSV file
data = pd.read_csv("population_growth_beta_euler.csv")
time = data["Time"]
population = data["Population"]

steady_state_population = 20  # Calculated Analytically

# Plot population over time
plt.figure(figsize=(10, 6))
plt.plot(time, population, label="Numerical Population", color="blue")
plt.axvline(x=100, color="green", linestyle="--", label="Switch to Beta (\u03B2) Dynamics at t=100")  # Vertical line at t=100
plt.axhline(y=steady_state_population, color="red", linestyle="--", label=f"Steady-State Population ({steady_state_population} million)")
plt.legend(fontsize=12)
plt.grid(True)
plt.title("Population Growth with Anomalous Death Rate", fontsize=16)
plt.xlabel("Time (years)", fontsize=14)
plt.ylabel("Population (millions)", fontsize=14)


plt.show()
```
