```python
import pandas as pd
import matplotlib.pyplot as plt

angles = [15, 25, 35, 45]

plt.figure(figsize=(10, 6)) 

# Plot the original angles
for angle in angles:
    filename = f"Trajectory_Of_{angle}_Degrees.csv"  # Match the C++ file naming
    data = pd.read_csv(filename)
    print(f"{angle}: Height range from {data['y'].max():.2f} m to {data['y'].min():.2f} m")
    plt.plot(data["x"], data["y"], label=f"{angle}°")

import glob
best_file = glob.glob("Best_Trajectory_37_Degrees.csv")
if best_file:
    best_file = best_file[0] 
    best_angle = best_file.split("_")[-2] 
    try:
        data = pd.read_csv(best_file)
        print(f"Best angle {best_angle}°: Height range from {data['y'].max():.2f} m to {data['y'].min():.2f} m")
        plt.plot(data["x"], data["y"], label=f"Best Angle: {best_angle}°", linestyle="dashed", linewidth=2)
    except FileNotFoundError:
        print(f"File {best_file} not found!")

# Graph labels and title
plt.xlabel("Distance (m)")
plt.ylabel("Height (m)")
plt.title("Projectile Trajectories")
plt.legend()
plt.grid(True)

# Show the plot
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Define angles used in the simulation
angles = [15, 25, 35, 33, 45]  # Angles used in the C++ program

# 🚀 Plot Projectile Trajectories
plt.figure(figsize=(10, 6))

for angle in angles:
    filename = f"Trajectory_With_Wind_Of_{angle}_Degrees.csv"
    try:
        data = pd.read_csv(filename)  # Read trajectory CSV
        if angle == 33:
            plt.plot(data["x"], data["y"], label=f"Best angle: 33.5°")  # Custom label for best angle
        else:
            plt.plot(data["x"], data["y"], label=f"{angle}°")  # Normal label for other angles

    except FileNotFoundError:
        print(f"⚠️ File {filename} not found! Make sure the C++ program generated it.")


# Graph labels and title
plt.xlabel("Distance (m)")
plt.ylabel("Height (m)")
plt.title("Projectile Trajectories with Headwind (25 mph)")
plt.legend()
plt.grid(True)

# Show the trajectory plot
plt.show()


# 📊 Plot Impact Speed vs. Launch Angle
impact_filename = "Impact_Speed_Wind.csv"

try:
    impact_data = pd.read_csv(impact_filename)  # Read impact speed CSV

    plt.figure(figsize=(10, 6))
    plt.plot(impact_data["theta"], impact_data["impact_speed"], marker="o", linestyle="", color="r", label="Impact Speed")

    # Labels and title
    plt.xlabel("Launch Angle (degrees)")
    plt.ylabel("Impact Speed (m/s)")
    plt.title("Impact Speed vs. Launch Angle (With Wind)")
    plt.legend()
    plt.grid(True)

    # Show the impact speed plot
    plt.show()

except FileNotFoundError:
    print(f"⚠️ File {impact_filename} not found! Make sure the C++ program generated it.")
```
