```python
import numpy as np
import matplotlib.pyplot as plt

# 1) Load the spectrum data
data = np.loadtxt("fourier_spectrum.dat")
freq = data[:, 0]
power = data[:, 1]

# 2) Plot power vs. frequency
plt.figure(figsize=(8, 5))
plt.plot(freq, power, 'b-', linewidth=1)
plt.title("Power Spectrum of y(x_s, t)")
plt.xlabel("Frequency (Hz)")
plt.ylabel("Power")

# 3) (Optional) Mark theoretical harmonic frequencies
#    For a string with length L=2.0, tension FT=720, M=0.025 => c = sqrt(FT/mu).
L = 2.0
FT = 720.0
M = 0.025
mu = M / L
c = np.sqrt(FT / mu)

f1 = c / (2.0 * L)  # fundamental frequency
n_harmonics = 15
for n in range(1, n_harmonics+1):
    f_h = n * f1
    plt.axvline(f_h, color='r', linestyle='--', alpha=0.7)

# 4) Optionally limit x-axis to Nyquist or some range
nyquist = freq[-1]  # naive DFT freq up to ~ (N-1)/(N dt)
plt.xlim(0, nyquist)

plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

data = np.loadtxt("fourier_spectrum.dat")
freq = data[:,0]
power = data[:,1]

N = len(freq)
# For a real signal, we only need the first half
freq_half = freq[:N//2]
power_half = power[:N//2]

plt.figure(figsize=(8,5))
plt.semilogy(freq_half, power_half, 'g-')  # semilog y-axis
plt.title("Power Spectrum")
plt.xlabel("Frequency (Hz)")
plt.ylabel("log Power")

# Mark harmonics
L = 2.0
FT = 720.0
M = 0.025
mu = M / L
c = np.sqrt(FT / mu)
f1 = c / (2.0 * L)
for n in range(1,(16*3)):
    f_h = n * f1
    plt.axvline(f_h, color='k', linestyle='--', alpha=0.5)

# Zoom to ~ 0..1kHz
plt.xlim(0, 3000)
plt.grid(True)
plt.tight_layout()
plt.show()
```
